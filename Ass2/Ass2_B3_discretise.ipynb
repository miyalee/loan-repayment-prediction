# 1B_3 Discretise
## Opening Data

```python
import numpy as np
import pandas as pd
import sklearn as sk
```

```python
data_path = "./32130_2026A_88.csv"
adata = pd.read_csv(data_path)
# adata
adata.info()
```

```python
# Young (ages 21 and younger)
# Early Adulthood (ages 22--34).
# early Middle Age (ages 35--44),
# late Middle Age (ages 45--64), and
# Late Adulthood (ages 65 and older).
bins = [0, 21, 34, 44, 64, 120]
names = ['Young', 'Early Adulthood', 'early Middle Age', 'late Middle Age', 'Late Adulthood']

adata['age_categorized'] = pd.cut(adata['age'], bins, labels=names, include_lowest=True)


print(adata[['age', 'age_categorized']].sort_values('age'))
adata[['age', 'age_categorized']].sort_values('age').groupby('age_categorized').count()
```

## Export CSV

```python
final=adata[['age', 'age_categorized']].sort_values('age')
final.to_csv('1B_3_Discretise.csv', index=False)
final
```
