# 1B_2 Normalise
## Opening Data

```python
import numpy as np
import pandas as pd
import sklearn as sk
```

```python
data_path = "./32130_2026A_88.csv"
adata = pd.read_csv(data_path)
# adata
adata.info()
```

## Min-Max

```python
# min-max transformation
scaler = sk.preprocessing.MinMaxScaler()

# fit and transform
adata['annual_income (min-max)'] = scaler.fit_transform(adata[['annual_income']])
print(adata[['annual_income', 'annual_income (min-max)']])
```

## Z-Score

```python
# z-score transformation
scaler = sk.preprocessing.StandardScaler()

# fit and transform
adata['annual_income (z-score)'] = scaler.fit_transform(adata[['annual_income']])
print(adata[['annual_income', 'annual_income (z-score)']])
```

## Export CSV

```python
final=adata[['annual_income', 'annual_income (min-max)', 'annual_income (z-score)']]
final.to_csv('1B_2_Normalise.csv', index=False)
final
```
