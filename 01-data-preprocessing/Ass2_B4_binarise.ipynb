# 1B_4 Binarise
## Opening Data

```python
import numpy as np
import pandas as pd
import sklearn as sk
```

```python
data_path = "./32130_2026A_88.csv"
adata = pd.read_csv(data_path)
# adata
adata.info()
```

```python
# Binarise the gender variable [with values "0" or "1"]
ranges = {'Male': 0, 'Female': 1, 'Other': 1}

print(adata.groupby('gender').size()) # 2% gender data is Other

adata['gender_binarised'] = adata['gender'].map(ranges)
print(adata[['gender', 'gender_binarised']].sort_values('gender'))
```

## Export CSV

```python
final=adata[['gender', 'gender_binarised']].sort_values('gender')
final.to_csv('1B_4_Binarise.csv', index=False)
final
```
